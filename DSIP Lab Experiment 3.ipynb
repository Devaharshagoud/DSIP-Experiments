```python
import numpy as np
import matplotlib.pyplot as plt
import os
```

```python
output_folder = "Correlation_Images"

os.makedirs(output_folder, exist_ok=True)

print("Folder Created Successfully")
```

```python
signal1 = np.array([1, 3, 2, 5, 4, 2, 1])

signal2 = np.array([2, 1, 4, 3, 5, 2, 1])

print("Signal 1 =", signal1)
print("Signal 2 =", signal2)
```

```python
plt.figure(figsize=(8,4))

plt.stem(signal1)

plt.title("Original Signal 1")

plt.xlabel("Sample")

plt.ylabel("Amplitude")

plt.grid(True)

plt.tight_layout()

plt.savefig(os.path.join(output_folder,"Original_Signal1.png"))

plt.show()
```

```python
plt.figure(figsize=(8,4))

plt.stem(signal2)

plt.title("Original Signal 2")

plt.xlabel("Sample")

plt.ylabel("Amplitude")

plt.grid(True)

plt.tight_layout()

plt.savefig(os.path.join(output_folder,"Original_Signal2.png"))

plt.show()
```

```python
auto1 = np.correlate(signal1, signal1, mode="full")

lags1 = np.arange(-(len(signal1)-1), len(signal1))

plt.figure(figsize=(8,4))

plt.stem(lags1, auto1)

plt.title("Autocorrelation of Signal 1")

plt.xlabel("Lag")

plt.ylabel("Correlation")

plt.grid(True)

plt.tight_layout()

plt.savefig(os.path.join(output_folder,"Autocorrelation_Signal1.png"))

plt.show()
```

```python
auto2 = np.correlate(signal2, signal2, mode="full")

lags2 = np.arange(-(len(signal2)-1), len(signal2))

plt.figure(figsize=(8,4))

plt.stem(lags2, auto2)

plt.title("Autocorrelation of Signal 2")

plt.xlabel("Lag")

plt.ylabel("Correlation")

plt.grid(True)

plt.tight_layout()

plt.savefig(os.path.join(output_folder,"Autocorrelation_Signal2.png"))

plt.show()
```

```python
cross12 = np.correlate(signal1, signal2, mode="full")

lags12 = np.arange(-(len(signal2)-1), len(signal1))

plt.figure(figsize=(8,4))

plt.stem(lags12, cross12)

plt.title("Cross Correlation of Signal 1 with Signal 2")

plt.xlabel("Lag")

plt.ylabel("Correlation")

plt.grid(True)

plt.tight_layout()

plt.savefig(
    os.path.join(
        output_folder,
        "CrossCorrelation_Signal1.png"
    )
)

plt.show()
```

```python
cross21 = np.correlate(signal2, signal1, mode="full")

lags21 = np.arange(-(len(signal1)-1), len(signal2))

plt.figure(figsize=(8,4))

plt.stem(lags21, cross21)

plt.title("Cross Correlation of Signal 2 with Signal 1")

plt.xlabel("Lag")

plt.ylabel("Correlation")

plt.grid(True)

plt.tight_layout()

plt.savefig(
    os.path.join(
        output_folder,
        "CrossCorrelation_Signal2.png"
    )
)

plt.show()
```

```python
plt.figure(figsize=(10,5))

plt.plot(lags1, auto1,
         marker='o',
         linewidth=2,
         label="Signal 1")

plt.plot(lags2, auto2,
         marker='s',
         linewidth=2,
         label="Signal 2")

plt.title("Comparison of Auto Correlation")

plt.xlabel("Lag")

plt.ylabel("Correlation")

plt.legend()

plt.grid(True)

plt.tight_layout()

plt.savefig(
    os.path.join(
        output_folder,
        "AutoCorrelation_Comparison.png"
    )
)

plt.show()
```

```python
plt.figure(figsize=(10,5))

plt.plot(lags12, cross12,
         marker='o',
         linewidth=2,
         label="Signal1 vs Signal2")

plt.plot(lags21, cross21,
         marker='s',
         linewidth=2,
         label="Signal2 vs Signal1")

plt.title("Comparison of Cross Correlation")

plt.xlabel("Lag")

plt.ylabel("Correlation")

plt.legend()

plt.grid(True)

plt.tight_layout()

plt.savefig(
    os.path.join(
        output_folder,
        "CrossCorrelation_Comparison.png"
    )
)

plt.show()
```
