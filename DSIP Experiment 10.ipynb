```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
imread = cv2.imread(r"C:\Users\devah\Documents\ICT MU\Sem 5\DSIP\Lab\Images\Krishna 4k.jpeg")

image = cv2.cvtColor(imread, cv2.COLOR_BGR2RGB)

plt.imshow(image)
plt.axis("off")
plt.title("Original Image")
plt.show()
```

```python
median_filtered = cv2.medianBlur(image, 5)

plt.imshow(median_filtered)
plt.axis("off")
plt.title("Median Filtered Image")
plt.show()
```

```python
bilateral_filtered = cv2.bilateralFilter(
    image,
    d=9,
    sigmaColor=75,
    sigmaSpace=75
)

plt.imshow(bilateral_filtered)
plt.axis("off")
plt.title("Bilateral Filtered Image")
plt.show()
```

```python
plt.figure(figsize=(15, 5))

plt.subplot(1, 3, 1)
plt.imshow(image)
plt.title("Original Image")
plt.axis("off")

plt.subplot(1, 3, 2)
plt.imshow(median_filtered)
plt.title("Median Filtered Image")
plt.axis("off")

plt.subplot(1, 3, 3)
plt.imshow(bilateral_filtered)
plt.title("Bilateral Filtered Image")
plt.axis("off")

plt.tight_layout()
plt.show()
```

```python
from PIL import Image
from PIL import ImageFilter
```

```python
image_pil = Image.open(
    r"C:\Users\devah\Documents\ICT MU\Sem 5\DSIP\Lab\Images\Krishna 4k.jpeg"
)
```

```python
min_filtered = image_pil.filter(ImageFilter.MinFilter(size=5))
```

```python
plt.imshow(min_filtered)
plt.title("Min Filter")
plt.axis("off")
plt.show()
```

```python
max_filtered = image_pil.filter(ImageFilter.MaxFilter(size=5))
```

```python
plt.imshow(max_filtered)
plt.title("Max Filter")
plt.axis("off")
plt.show()
```

```python
plt.figure(figsize=(10, 5))

plt.subplot(1, 2, 1)
plt.imshow(min_filtered)
plt.title("Min Filter")
plt.axis("off")

plt.subplot(1, 2, 2)
plt.imshow(max_filtered)
plt.title("Max Filter")
plt.axis("off")

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(15, 8))

plt.subplot(2, 3, 1)
plt.imshow(image)
plt.title("Original Image")
plt.axis("off")

plt.subplot(2, 3, 2)
plt.imshow(median_filtered)
plt.title("Median Filter")
plt.axis("off")

plt.subplot(2, 3, 3)
plt.imshow(bilateral_filtered)
plt.title("Bilateral Filter")
plt.axis("off")

plt.subplot(2, 3, 5)
plt.imshow(min_filtered)
plt.title("Min Filter")
plt.axis("off")

plt.subplot(2, 3, 6)
plt.imshow(max_filtered)
plt.title("Max Filter")
plt.axis("off")

plt.tight_layout()
plt.show()
```
